```python
n=17
isPrime = True #假設n是質數
for i in range(2,n):
    if n%i==0:
        print(f'{n}不是質數,因為它可以被{i}除盡')
        isPrime = False #確定n不是質數
        break #只要數字能被除盡，就跳離迴圈

if isPrime == True:#判斷我們的假設是否還成立
    print(f'{n}是質數')
```

# 練習
將上面每一row最後面的那個逗號移除。

提示:用if指令

```python
for i in range(1,6):
    for j in range(1,6):
        print(f'{i}*{j}={i*j}',end=',')
    print()
```

```python
for j in range(1,6):
    print(f'1*{j}={1*j}',end=',')
print()

for j in range(1,6):
    print(f'2*{j}={2*j}',end=',')
print()

for j in range(1,6):
    print(f'3*{j}={3*j}',end=',')
print()
```
